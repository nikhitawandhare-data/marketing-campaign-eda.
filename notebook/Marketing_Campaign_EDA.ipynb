```python
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
```

```python
outliers = df[
    (df["Income"] < lower_limit) |
    (df["Income"] > upper_limit)
]

print("Number of outliers:", len(outliers))

outliers[["ID", "Income", "Country"]].sort_values(
    by="Income", ascending=False
)
```

```python
age_bins = [17, 25, 35, 45, 55, 100]

age_labels = [
    "18-25",
    "26-35",
    "36-45",
    "46-55",
    "56-100"
]

df["Age_Group"] = pd.cut(
    df["Age"],
    bins=age_bins,
    labels=age_labels
)

print(df["Age_Group"].value_counts().sort_index())
```

## Customer Age Group Segmentation

Customers were divided into five age groups:

| Age Group | Customer Count |
|---|---:|
| 18-25 | 91 |
| 26-35 | 386 |
| 36-45 | 758 |
| 46-55 | 486 |
| 56-100 | 516 |

### Key Findings

- The largest age group is 36-45, with 758 customers.
- The smallest age group is 18-25, with 91 customers.
- Three customers above 100 were not assigned to an age group.

### Business Observation

The customer dataset is concentrated in the 36-45 age group.
Further analysis is needed to understand spending and campaign
response patterns across age groups.

```python
age_campaign_analysis = df.groupby(
    "Age_Group",
    observed=True
).agg(
    Total_Customers=("ID", "count"),
    Campaign_Accepted=("Response", "sum"),
    Acceptance_Rate=("Response", "mean")
)

age_campaign_analysis["Acceptance_Rate"] *= 100

age_campaign_analysis["Acceptance_Rate"] = (
    age_campaign_analysis["Acceptance_Rate"].round(2)
)

print(age_campaign_analysis)
```

## Campaign Acceptance by Customer Age Group

| 36-45 | 758 | 109 | 14.38% |
| 46-55 | 486 | 59 | 12.14% |
| 56-100 | 516 | 79 | 15.31% |

### Key Findings

- The 26-35 age group has the highest acceptance rate (18.65%).
- The 46-55 age group has the lowest acceptance rate (12.14%).
- The 36-45 age group contributes the highest number of acceptances (109).

### Business Recommendation

Investigate customer preferences in the 26-35 age group

Further analysis is needed before drawing causal conclusions.

```python
age_spending_analysis = df.groupby(
    "Age_Group",
    observed=True
).agg(
    Total_Customers=("ID", "count"),
    Total_Spending=("Total_Spending", "sum"),
    Average_Spending=("Total_Spending", "mean")
)

age_spending_analysis["Average_Spending"] = (
    age_spending_analysis["Average_Spending"].round(2)
)

print(age_spending_analysis)
```

```python
product_columns = [
    "MntWines",
    "MntFruits",
    "MntMeatProducts",
    "MntFishProducts",
    "MntSweetProducts",
    "MntGoldProds"
]

age_product_analysis = df.groupby(
    "Age_Group",
    observed=True
)[product_columns].mean().round(2)

print(age_product_analysis)
```

```python
wine_campaign_analysis = df.groupby("Response")["MntWines"].agg(
    Total_Customers="count",
    Average_Wine_Spending="mean",
    Median_Wine_Spending="median",
    Total_Wine_Spending="sum"
)

print(wine_campaign_analysis.round(2))
```

### Key Findings: Wine Spending by Campaign Response

- 334 customers accepted the latest campaign.
- 1,906 customers did not accept it.
- Campaign acceptors had average wine spending of 502.70.
- Non-acceptors had average wine spending of 269.10.
- Average wine spending was approximately 86.8% higher among campaign acceptors.
- Median wine spending was also higher among campaign acceptors.

Business Insight:
Customers who accepted the latest campaign showed higher average wine spending. The marketing team should investigate whether existing wine preferences influenced campaign acceptance.

Limitation:
This analysis identifies an association and does not prove that the campaign caused higher wine spending.
